```python
from sklearn.datasets import load_diabetes
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
model=DecisionTreeRegressor(random_state=42)

df=load_diabetes(as_frame=True).frame
x=df.drop("target",axis=1)
y=df["target"]

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
model=DecisionTreeRegressor()
model.fit(x_train,y_train)
```

```python
from sklearn.metrics import r2_score,mean_squared_error

y_pred_train=model.predict(x_train)
y_pred_test=model.predict(x_test)

print("MSE for TRAINING",mean_squared_error(y_train,y_pred_train))
print("MSE for TESTING",mean_squared_error(y_test,y_pred_test))

print("R2_SCORE for TRAINING",r2_score(y_train,y_pred_train))
print("R2_SCORE for TESTING",r2_score(y_test,y_pred_test))
```

```python
# MSE AND R2_SCORE shows overfitting as we predicted on train data
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
plt.figure(figsize=(10,8))
plot_tree(
    model,
    feature_names=x.columns,
    filled=True)
plt.tight_layout()
```

```python
# lets do pruning
model=DecisionTreeRegressor(random_state=42,max_depth=7)

df=load_diabetes(as_frame=True).frame
x=df.drop("target",axis=1)
y=df["target"]

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
model.fit(x_train,y_train)
```

```python
from sklearn.metrics import r2_score,mean_squared_error

y_pred_train=model.predict(x_train)
y_pred_test=model.predict(x_test)

print("MSE for TRAINING",mean_squared_error(y_train,y_pred_train))
print("MSE for TESTING",mean_squared_error(y_test,y_pred_test))

print("R2_SCORE for TRAINING",r2_score(y_train,y_pred_train))
print("R2_SCORE for TESTING",r2_score(y_test,y_pred_test))
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
plt.figure(figsize=(10,8))
plot_tree(
    model,
    feature_names=x.columns,
    filled=True)
plt.tight_layout()
```
