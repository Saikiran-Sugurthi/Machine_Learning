```python
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
import seaborn as sns
```

```python
x,y=make_moons(
    n_samples=300,
    noise=0.05,
    random_state=42)
#Scaling
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
sns.scatterplot(x=x_scaled[:,0],y=x_scaled[:,1])
```

# LETS TEST WITH KMEANS

```python
from sklearn.cluster import KMeans
kmeans=KMeans(2)
labels=kmeans.fit_predict(x_scaled)
sns.scatterplot(x=x_scaled[:,0],y=x_scaled[:,1],c=labels)

## see it is drawing a line in bw ..and producing wrong clusters
```

# Now lets test with DBSCAN

```python
from sklearn.cluster import DBSCAN
db=DBSCAN(
    eps=0.5,
    min_samples=5)
lab=db.fit_predict(x_scaled)
sns.scatterplot(x=x_scaled[:,0],y=x_scaled[:,1],c=lab)
```
